```python
%cd /sphenix/user/hwyu/calotrack_tree/scripts

import torch
from torchmetrics.clustering import AdjustedRandScore
from util import *

# Initialize the metric
ari = AdjustedRandScore()

# Example predictions and ground truth labels
target = torch.tensor([-1, -1, 1, 1, 2, 2])
preds1 = torch.tensor([1, 2, 3, 3, 4, 4])
preds2 = torch.tensor([1, 2, 4, 4, 3, 3])

# Compute ARI
score1 = ari(preds1, target)
score2 = ari(preds2, target)
print(f"ARI 1: {score1}")
print(f"ARI 2: {score2}")
```

```python
groups = [[0, 1], [2, 3], [2, 5]]
groupids = get_group_ids(groups, 6)
print(f"Group IDs: {groupids}")
```

```python
from sklearn.metrics import adjusted_rand_score

# Convert torch tensors to numpy arrays for sklearn
target_np = target.numpy()
preds1_np = preds1.numpy()
preds2_np = preds2.numpy()

# Calculate ARI using sklearn
sklearn_ari1 = adjusted_rand_score(target_np, preds1_np)
sklearn_ari2 = adjusted_rand_score(target_np, preds2_np)

# Convert groupids to a tensor and calculate ARI
groupids_tensor = torch.tensor(groupids)
sklearn_ari_groups = adjusted_rand_score(target_np, groupids_tensor.numpy())

print(f"Sklearn ARI 1: {sklearn_ari1}")
print(f"Sklearn ARI 2: {sklearn_ari2}")
print(f"Sklearn ARI for groupids: {sklearn_ari_groups}")
```
